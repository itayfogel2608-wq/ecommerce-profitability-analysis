```python
'''
[
  {"product_id": 101, "name": "Wireless Mouse", "category": "Electronics", "cost_price": 40, "sell_price": 90, "quantity_sold": 150},
  {"product_id": 102, "name": "Desk Lamp", "category": "Home", "cost_price": 60, "sell_price": 120, "quantity_sold": 85},
  {"product_id": 103, "name": "Mechanical Keyboard", "category": "Electronics", "cost_price": 150, "sell_price": 300, "quantity_sold": 40},
  {"product_id": 104, "name": "Coffee Mug", "category": "Home", "cost_price": 10, "sell_price": 35, "quantity_sold": 300},
  {"product_id": 105, "name": "Laptop Stand", "category": "Accessories", "cost_price": 45, "sell_price": 110, "quantity_sold": 120}
]


'''
import json
import openpyxl 
from openpyxl.styles import Font,PatternFill,Alignment,Border,Side 

json_data = [
  {"product_id": 101, "name": "Wireless Mouse", "category": "Electronics", "cost_price": 40, "sell_price": 90, "quantity_sold": 150},
  {"product_id": 102, "name": "Desk Lamp", "category": "Home", "cost_price": 60, "sell_price": 120, "quantity_sold": 85},
  {"product_id": 103, "name": "Mechanical Keyboard", "category": "Electronics", "cost_price": 150, "sell_price": 300, "quantity_sold": 40},
  {"product_id": 104, "name": "Coffee Mug", "category": "Home", "cost_price": 10, "sell_price": 35, "quantity_sold": 300},
  {"product_id": 105, "name": "Laptop Stand", "category": "Accessories", "cost_price": 45, "sell_price": 110, "quantity_sold": 120}
]

# 1. כאן אתה מדביק את הנתיב המדויק שלך (שים לב ל-r בהתחלה ולשם הקובץ בסוף)
exact_json_path = r"C:\Users\itayf\OneDrive\Desktop\ֵתואר\פרוייקטים עצמאיים ללינקדין\store_data\store_data.json"

print('Creating JSON file in the exact folder...')

# 2. שמירה של הקובץ ישירות לנתיב שהגדרנו
with open(exact_json_path, 'w', encoding='utf-8') as f:
    json.dump(json_data, f, ensure_ascii=False, indent=4)
    
print("JSON saved successfully!")

def jsonToExcelMultiSheet(json_file,excel_file):
    print(f'reading data from{json_file}')
    with open(json_file,'r',encoding='utf-8') as f:
        data=json.load(f)
    print(f'creating Excel Workbook...')
    wb=openpyxl.Workbook()
    ws=wb.active
  
    thin_border=Border(
        left=Side(style='thin'),
        right=Side(style='thin'),
        top=Side(style='thin'),
        bottom=Side(style='thin')
    )


    ws.title='sales Report'
    ws.merge_cells('A1:H1')
    ws['A1']=ws.title
    ws['A1'].font=Font(size=16,bold=True)
    ws['A1'].alignment=Alignment(horizontal='center')

    headers=['product_id','name','category','cost_price','sell_price','quantity_sold','revenue','profit','profit margin']
    for col,header in enumerate(headers,start=1):
        cell=ws.cell(row=3,column=col)
        cell.value=header
        cell.font=Font(bold=True,color='FFFFFF')
        cell.fill=PatternFill(start_color='4472C4',fill_type='solid')
        cell.alignment=Alignment(horizontal='center')
    row=4
    for product in data:
        revenue=product['sell_price']*product['quantity_sold']
        profit_per_unit=product['sell_price']-product['cost_price']
        profit=profit_per_unit*product['quantity_sold']
        profit_margin=profit/revenue
        ws.cell(row=row,column=1).value=product['product_id']
        ws.cell(row=row,column=2).value=product['name']
        ws.cell(row=row,column=3).value=product['category']
        ws.cell(row=row,column=4).value=product['cost_price']
        ws.cell(row=row,column=5).value=product['sell_price']
        ws.cell(row=row,column=6).value=product['quantity_sold']
        ws.cell(row=row,column=7).value=revenue 
        ws.cell(row=row,column=8).value=profit 
        ws.cell(row=row,column=9).value=profit_margin 

        quantity_cell=ws.cell(row=row,column=6)
        if product['quantity_sold']>100:
            quantity_cell.fill=PatternFill(start_color='FF6B6B',fill_type='solid')
        elif product['quantity_sold']>=60 and product['quantity_sold']<100:
            quantity_cell.fill=PatternFill(start_color='FFD93D',fill_type='solid')
        else:
            quantity_cell.fill=PatternFill(start_color='6BCF7F',fill_type='solid')
        row+=1
    for r in range(3,row):
        for c in range(1,10):
            ws.cell(row=r,column=c).border=thin_border
    for col_letter in ['A','B','C','D','E','F','G','H','I']:
        ws.column_dimensions[col_letter].width=20

jsonToExcelMultiSheet('store_data.json','store_data.xlsx')
    




```
